## Valores y tipos

Todo dato que maneja un programa es un **valor**, y cada valor tiene un **tipo**. En este tema usamos cuatro:

| Tipo | Qué guarda | Ejemplos |
|---|---|---|
| `int` | números enteros | `17`, `-3`, `2026` |
| `float` | números con decimales | `1.65`, `-0.5`, `3.0` |
| `str` | texto (una «cadena» de caracteres), entre comillas | `"Sucre"`, `'hola'`, `"42"` |
| `bool` | verdadero o falso | `True`, `False` |

Ojo con los decimales: en Python se escriben con **punto**, no con coma. Una estatura de 1,65 m se escribe `1.65`. La función `type` te dice de qué tipo es un valor:

````python
print(type(17))
print(type(1.65))
print(type("Sucre"))
print(type(True))
print(type("42"))
````

El último es un detalle importante: `"42"` está entre comillas, así que para Python es **texto**, no un número. Con los números se puede hacer aritmética:

````python
print(7 + 2)
print(7 - 2)
print(7 * 2)
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(7 ** 2)
````

- `*` multiplica y `**` eleva a una potencia: `7 ** 2` es 7².
- `/` divide y **siempre** da un `float`, aunque la división sea exacta: `6 / 2` da `3.0`.
- `//` es la división entera (cuántas veces cabe) y `%` es el **resto**. 7 entre 2 da 3 y sobra 1. El resto sirve mucho: un número es par cuando `n % 2` vale 0.

Como en matemáticas, primero se calculan las potencias, después las multiplicaciones y divisiones, y al final las sumas y restas. Si quieres otro orden, usa paréntesis. El texto también tiene sus operaciones: `+` pega dos textos y `*` repite uno. `len` cuenta sus caracteres.

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
print("Hola" + " " + "mundo")
print("ja" * 3)
print(len("Cochabamba"))
````

## Variables: cajas con nombre

Una **variable** es un nombre que guarda un valor, como la caja de la actividad. Se crea con el signo `=`:

````python
edad = 15
nombre = "Ana"
print(nombre, "tiene", edad, "años")

edad = edad + 1
print(nombre, "ahora tiene", edad, "años")
````

En Python, `=` **no** significa «es igual a» como en matemáticas: significa «calcula lo de la derecha y guárdalo en la caja de la izquierda». Por eso `edad = edad + 1` tiene sentido: toma el valor actual de `edad` (15), le suma 1 y guarda el 16 en la misma caja. El 15 se pierde.

Cuando a `print` le das varias cosas separadas por comas, las muestra en una línea con un espacio entre cada una. Algunas reglas para los nombres:

- Pueden tener letras, números y guion bajo (`nota_final`, `edad2`), pero no pueden empezar con un número ni tener espacios.
- Mayúsculas y minúsculas cuentan: `edad` y `Edad` son dos cajas distintas.
- Elige nombres que digan qué guardan: `precio` se entiende mejor que `p`. Es costumbre escribirlos en minúsculas y sin tildes.

## Juntar texto y números: f-strings

Para armar un mensaje con valores adentro, lo más cómodo es una **f-string**: un texto con una `f` antes de las comillas, donde lo que va entre llaves se calcula y se reemplaza por su valor.

````python
nombre = "Ana"
edad = 16
print(f"{nombre} tiene {edad} años.")
print(f"El próximo año tendrá {edad + 1}.")

pi = 3.14159
print(f"Pi con dos decimales: {pi:.2f}")
````

El `:.2f` después del valor quiere decir «muéstralo con 2 decimales». ¿Y si intentas pegar texto y número con `+`?

*La celda que sigue da un error a propósito: es parte de la explicación.*

````python
edad = 16
print("Tengo " + edad + " años")
````

Es tu primer **error**. La última línea dice qué pasó: un `TypeError` (error de tipo), porque solo se puede «concatenar» (pegar) un `str` con otro `str`, no con un `int`. La flecha señala la línea 2 de la celda. Hay dos arreglos: usar una f-string, o convertir el número en texto con `str`. Las funciones `int`, `float` y `str` convierten de un tipo a otro:

````python
print("Tengo " + str(edad) + " años")
print(int("42") + 1)
print(float("1.65") * 2)
````

## Decisiones: if, elif y else

Un programa que siempre hace lo mismo sirve de poco. Para decidir, primero hay que **comparar**. Cada comparación da un `bool`, `True` o `False`:

| Escribes | Significa |
|---|---|
| `a == b` | ¿son iguales? (dos signos `=`: uno solo es guardar) |
| `a != b` | ¿son distintos? |
| `a > b`, `a < b` | ¿mayor?, ¿menor? |
| `a >= b`, `a <= b` | ¿mayor o igual?, ¿menor o igual? |

Las comparaciones se combinan con `and` (las dos tienen que cumplirse), `or` (basta con una) y `not` (lo contrario). Supongamos que en un curso se aprueba con 51 sobre 100:

````python
nota = 58
print(nota >= 51)
print(nota == 100)
print(nota != 100)
print(nota >= 51 and nota < 70)
print(not nota >= 51)
````

Con `if` (si), el programa ejecuta un bloque de líneas solo cuando la condición es `True`; con `else` (si no), ejecuta otro bloque cuando es `False`:

````python
nota = 58
if nota >= 51:
    print("Aprobaste")
else:
    print("Todavía no apruebas")
print("Fin")
````

Mira la forma: la condición termina con **dos puntos**, y las líneas que dependen de ella van **sangradas**, es decir, corridas 4 espacios a la derecha (Colab lo hace solo al pulsar Enter después de los dos puntos). La sangría no es decoración: es la manera en que Python sabe qué líneas están dentro del `if`. `print("Fin")` no tiene sangría, así que se ejecuta siempre. Si olvidas la sangría, Python se queja:

*La celda que sigue da un error a propósito: es parte de la explicación.*

````python
if nota >= 51:
print("Aprobaste")
````

Cuando hay más de dos caminos, se usa `elif` («si no, si…»). Python revisa las condiciones **de arriba abajo** y ejecuta solo el bloque de la **primera** que se cumple. Así, la calculadora de IMC puede dar una categoría:

````python
if imc < 18.5:
    categoria = "bajo peso"
elif imc < 25:
    categoria = "peso normal"
elif imc < 30:
    categoria = "sobrepeso"
else:
    categoria = "obesidad"
print(f"IMC {imc:.1f}: {categoria}")
````

En el segundo `elif` no hace falta escribir `imc >= 18.5 and imc < 25`: si el programa llegó ahí, ya sabe que `imc < 18.5` fue falso. Un aviso importante: estos cortes son los que se usan para **adultos**. En adolescentes que todavía están creciendo, el IMC se compara con tablas según la edad y el sexo, así que este programa es un ejercicio de código, no un diagnóstico.

## Repetir: el bucle for

Para repetir algo, se usa un **bucle**. El bucle `for` recorre una serie de valores y ejecuta su bloque una vez por cada uno. `range(5)` da los números del 0 al 4: empieza en 0 y **no incluye** el 5.

````python
for i in range(5):
    print("Vuelta", i)
````

`range` puede recibir dónde empezar, dónde parar (sin incluirlo) y de cuánto en cuánto avanzar. Para ver todos sus números de una vez, conviértelo en una lista con `list`:

````python
print(list(range(1, 6)))
print(list(range(0, 20, 5)))
print(list(range(10, 0, -3)))
````

Con un `for` y una f-string, la tabla de multiplicar sale en tres líneas:

````python
numero = 7
for i in range(1, 11):
    print(f"{numero} x {i} = {numero * i}")
````

### Acumuladores: sumar dentro de un bucle

Volvamos a la actividad. Para sumar muchos números se usa un **acumulador**: una variable que empieza en 0 y a la que, en cada vuelta, se le suma algo. Este código es el programa de la actividad, con una línea más que imprime el valor de las variables en cada vuelta. Así puedes ver por dentro cualquier bucle:

````python
total = 0
print("   n  total")
for n in range(1, 5):
    total = total + n
    print(f"{n:>4} {total:>6}")
print("Suma:", total)
````

El `:>4` dentro de las llaves alinea el número a la derecha en un espacio de 4 caracteres, para que la tabla quede en columnas. Y ahora sí, la respuesta a la pregunta del principio:

````python
total = 0
for n in range(1, 101):
    total = total + n
print(total)
````

## Repetir mientras: el bucle while

A veces no sabes cuántas vueltas hacen falta. El bucle `while` (mientras) repite su bloque **mientras** la condición sea `True`, y la revisa antes de cada vuelta. Si ahorras 15 bolivianos por semana, ¿cuántas semanas tardas en juntar 100?

````python
ahorro = 0
semanas = 0
while ahorro < 100:
    ahorro = ahorro + 15
    semanas = semanas + 1
print(f"En {semanas} semanas juntas {ahorro} Bs.")
````

Algo dentro del bloque tiene que acercar la condición a `False`. Si olvidas la línea `ahorro = ahorro + 15`, el ahorro nunca cambia, la condición es `True` para siempre y el programa no termina: es un **bucle infinito**. En Colab lo detienes con el botón de detener de la celda (el mismo de reproducir, que mientras corre se convierte en un cuadrado) o con **Entorno de ejecución → Interrumpir ejecución**.

## Un juego: adivina el número

El módulo `random` (aleatorio) viene con Python y sirve para sortear números. Un **módulo** es un conjunto de herramientas que se activa con `import`. `random.randint(1, 20)` da un número entero al azar entre 1 y 20, **incluidos** los dos.

Si cada vez que corres el programa sale un número distinto, nadie puede comprobar sus resultados. Por eso, en la ruta, todo programa con azar fija una **semilla** con `random.seed`: con la misma semilla, los «sorteos» salen siempre iguales. Para jugar de verdad, borra esa línea.

El juego: la computadora piensa un número y tú intentas adivinarlo; después de cada intento te dice si el secreto es más alto o más bajo. Aquí las respuestas del jugador están en una **lista**, varios valores entre corchetes separados por comas (en el tema 2 vas a trabajar mucho con ellas). `intentos_jugador[0]` es el primer valor de la lista, `intentos_jugador[1]` el segundo, y así: se cuenta desde 0.

````python
import random

random.seed(7)
secreto = random.randint(1, 20)
# Lo que contesta el jugador, en orden
intentos_jugador = [10, 15, 12, 11]

intentos = 0
adivinado = False
while not adivinado:
    intento = intentos_jugador[intentos]
    intentos = intentos + 1
    if intento < secreto:
        print(f"{intento}: el secreto es más alto")
    elif intento > secreto:
        print(f"{intento}: el secreto es más bajo")
    else:
        print(f"{intento}: ¡correcto, en {intentos} intentos!")
        adivinado = True
````

`adivinado` es un `bool` que hace de interruptor: empieza en `False`, y el bucle sigue mientras **no** se haya adivinado. En Colab, para jugar tú, cambia la línea `intento = intentos_jugador[intentos]` por esta, y borra `random.seed(7)` y la lista:


## Una regla hecha a mano

Con lo que ya sabes puedes escribir un programa que «decide» como una persona. Este da un consejo de ropa según la temperatura de cada día. `for` también recorre una lista: en cada vuelta, `t` toma el siguiente valor de la lista, primero 3, después 11, y así hasta el último (lo verás a fondo en el tema 2):

````python
temperaturas = [3, 11, 17, 24, 8]
for t in temperaturas:
    if t < 10:
        consejo = "lleva chamarra"
    elif t < 20:
        consejo = "lleva una chompa"
    else:
        consejo = "ropa ligera"
    print(f"{t:>2} °C: {consejo}")
````

Funciona, pero hazte una pregunta: ¿quién decidió que el corte va en 10 grados y no en 12? Una persona, a ojo. Este programa es un **clasificador**: recibe un dato y lo pone en una categoría. Sus reglas las escribió alguien que pensó en el problema.

La inteligencia artificial que vas a aprender en esta ruta da vuelta esa idea: en lugar de escribir las reglas, le das a la computadora **muchos ejemplos** con su respuesta correcta y ella encuentra sola los números de la regla. En el tema 6 eso va a ser una recta que se ajusta a los datos; en el tema 11, un árbol de decisión que elige solo sus cortes, como el «menos de 10 grados» de aquí. Pero por dentro, todo eso sigue siendo variables, decisiones y bucles.

### 5. FizzBuzz

Escribe un programa que recorra los números del 1 al 15 y, para cada uno, imprima:

- `FizzBuzz` si es múltiplo de 3 y también de 5;
- `Fizz` si es múltiplo de 3 (pero no de 5);
- `Buzz` si es múltiplo de 5 (pero no de 3);
- el número mismo en cualquier otro caso.

Pista: un número es múltiplo de 3 cuando el resto de dividirlo entre 3 es 0.

````python
# @title Respuesta: código (doble clic para verlo)
for n in range(1, 16):
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")
    elif n % 3 == 0:
        print("Fizz")
    elif n % 5 == 0:
        print("Buzz")
    else:
        print(n)
````

<details>
<summary>Respuesta</summary>

El orden importa. Si la primera condición fuera `n % 3 == 0`, el 15 entraría ahí, imprimiría `Fizz` y nunca llegaría a la de `FizzBuzz`: el `if` ejecuta solo el **primer** bloque cuya condición se cumple. Por eso el caso más exigente va primero.

</details>

### 6. Doblar una hoja

Una hoja de papel tiene un grosor de 0,1 mm. Cada vez que la doblas por la mitad, el grosor se duplica. Usa un `while` para averiguar cuántas veces habría que doblarla para que el grosor pase de 1 metro (1000 mm), y cuánto mediría. (En la práctica, una hoja no se deja doblar más de unas pocas veces; aquí solo hacemos la cuenta.)

````python
# @title Respuesta: código (doble clic para verlo)
grosor = 0.1      # en milímetros
dobleces = 0
while grosor <= 1000:
    grosor = grosor * 2
    dobleces = dobleces + 1
print(f"{dobleces} dobleces: {grosor:.1f} mm")
````

<details>
<summary>Respuesta</summary>

No sabíamos de antemano cuántas vueltas hacían falta: para eso está `while`. Duplicar muchas veces crece muy rápido: con solo 14 dobleces se pasa del metro y medio.

</details>

## Errores comunes

- **Olvidar los dos puntos** al final de una línea con `if`, `elif`, `else`, `for` o `while`. El mensaje es `SyntaxError: expected ':'`.
- **Sangrías desparejas.** Las líneas de un mismo bloque deben tener exactamente la misma sangría. Usa siempre 4 espacios (Colab los pone solo).
- **Confundir `=` con `==`.** Un `=` guarda un valor; dos `==` comparan. `if nota = 51:` es un `SyntaxError`.
- **Escribir decimales con coma.** `1,65` no es un número en Python: es un par de valores. Se escribe `1.65`.
- **Hacer cuentas con lo que devuelve `input`.** Siempre es texto. Conviértelo con `int` o `float` antes de sumar o comparar con números.
- **Olvidar que `range` no incluye el final.** `range(1, 10)` llega hasta el 9. Para llegar al 10, escribe `range(1, 11)`.
- **Reiniciar el acumulador dentro del bucle.** Si `total = 0` está dentro del `for`, vuelve a 0 en cada vuelta y al final solo queda el último número.
- **Un `while` que nunca termina.** Algo en el bloque tiene que acercar la condición a `False`. Si la celda no termina, detenla y revisa qué variable debería cambiar.
- **Ejecutar las celdas en desorden.** Si una variable tiene un valor raro o aparece un `NameError` que no entiendes, reinicia la sesión y usa **Ejecutar todas**.
- **Leer el mensaje de error desde arriba.** Empieza por la última línea: ahí está el tipo de error y la explicación.
